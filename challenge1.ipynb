```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc_a = QuantumCircuit(1, 1)
qc_a.x(0)
qc_a.measure(0, 0)
qc_a.draw("mpl")
simulator = AerSimulator()
result = simulator.run(qc_a, shots=100).result()

print(result.get_counts())
qc_a.draw("mpl")
```

```python
qc_b = QuantumCircuit(1, 1)

qc_b.x(0)          # Apply the first X gate
qc_b.x(0)          # Apply the second X gate
qc_b.measure(0, 0) # Measure the qubit
simulator = AerSimulator()

result = simulator.run(qc_b, shots=100).result()

print(result.get_counts())
qc_b.draw("mpl")
```

```python
# Circuit : One H gate
qc_c = QuantumCircuit(1, 1)
qc_c.h(0)
qc_c.measure(0, 0)
simulator = AerSimulator()

result = simulator.run(qc_c, shots=100).result()

print(result.get_counts())
qc_c.draw("mpl")
```

```python
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
simulator = AerSimulator()

result = simulator.run(
    [qc_a, qc_b, qc_c],
    shots=100
).result()


counts_a = result.get_counts(0)
counts_b = result.get_counts(1)
counts_c = result.get_counts(2)


print("Circuit A results:", counts_a)
print("Circuit B results:", counts_b)
print("Circuit C results:", counts_c)

# Display the measurement histogram
FIG=plot_histogram(
    [counts_a, counts_b, counts_c],
    legend=["Circuit A: X", "Circuit B: X X", "Circuit C: H"],
    title="Measurement Results for 100 Shots"
)


display(FIG)
```
